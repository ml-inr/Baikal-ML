```python
import json, h5py

with open('data_manager/nu_classifier_ds_builder/testds_parts.json') as f:
    parts = json.load(f)

with h5py.File('data_manager/data/h5datasets/baikal_mc_merged.h5', 'r') as h5:
    for short, full in [('nuatm','nuatm_2020'), ('nue2','nue2_2020'), ('muatm', 'muatm_2020')]:
        available = set(int(k.split('_')[1]) for k in h5[full]['raw/ev_starts'].keys())
        requested = set(parts[short])
        missing = requested - available
        print(f'{short}: {len(requested)} requested, {len(missing)} missing: {sorted(missing)[:10]}')
```
